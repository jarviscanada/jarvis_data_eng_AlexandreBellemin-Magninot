## Load Data from CSV into DataFrame

The table has been imported using Databricks GUI. In my case, the table is in the jarvis_training catalogue, default schema, named retail.

```python
import pyspark.sql.functions as F
df = spark.table("jarvis_training.default.retail")
df.printSchema()
display(df.describe())
display(df)
```

## Total Invoice Amount Distribution
#### Calculate the invoice amount. 
Note: an invoice consists of one or more items where each item is a row in the df. 

```python
invoice_line_amount_df = (
    df
    .withColumn(
        "line_amount", 
        F.col("quantity") * F.col("unit_price") # Calculate the amount of each row
    )
)

invoice_amount_df = (
    invoice_line_amount_df
    .groupBy("invoice_no").agg(F.sum("line_amount").alias("amount")) # Calculate the amount of each invoices
    .filter(F.col("amount") > 0) # Keep only invoice placed, not cancelled
)
display(invoice_amount_df.describe())
display(invoice_amount_df)
```

#### Draw the distribution of invoice amount 
With min, max, median, mod, and mean, without managing outliers

```python
def show_distribution(df, col):
    stats = df.agg(
        F.min(col).alias("min"),
        F.max(col).alias("max"),
        F.mean(col).alias("mean"),
        F.median(col).alias("median"),
        F.mode(col).alias("mode")
    ).collect()[0]

    print(
        f"Minimum: {stats['min']:.2f}\n"
        f"Mean: {stats['mean']:.2f}\n"
        f"Median: {stats['median']:.2f}\n"
        f"Mode: {stats['mode']:.2f}\n"
        f"Maximum: {stats['max']:.2f}\n"
    )
```

```python
show_distribution(invoice_amount_df, "amount")
display(invoice_amount_df)
```

#### Draw the distribution for the first 85 quantiles of the invoice amount data 
With min, max, median, mod, and mean, and with outliers removed

```python
q85 = invoice_amount_df.approxQuantile("amount", [0.85], 0)[0]
remove_outliers = invoice_amount_df.filter(F.col("amount") <= q85)
show_distribution(remove_outliers, "amount")
display(remove_outliers)
```

## Monthly Placed and Canceled Orders
#### Create a new integer column with YYYYMM format

```python
year_month_df = df.withColumn("year_month", F.date_format(F.col("invoice_date"), "yyyyMM"))
display(year_month_df)
```

#### Create a dataframe with placed and canceled orders
Plot monthly placed and canceled orders

```python
orders_df = (
    year_month_df.groupBy("year_month")
    .agg(
        F.countDistinct(F.when(F.col("invoice_no").contains('C'), F.col("invoice_no"))).alias("canceled_orders"),
        (
            F.countDistinct(F.when(~F.col("invoice_no").contains('C'), F.col("invoice_no"))) -
            F.countDistinct(F.when(F.col("invoice_no").contains('C'), F.col("invoice_no")))
        ).alias("placed_orders")
    )
    .orderBy("year_month")
)
display(orders_df)
```

## Monthly Sales


```python
monthly_sales_df = (
    year_month_df
    .withColumn("amount", F.col("quantity") * F.col("unit_price"))
    .groupBy("year_month")
    .agg(F.sum("amount").alias("sales"))
    .orderBy("year_month")
)
display(monthly_sales_df)
```

##Monthly Sales Growth

```python
from pyspark.sql import Window

window_sales = Window.partitionBy(F.lit(1)).orderBy("year_month") # Explicit partitionBy to say to PySpark that we have only 1 serie

monthly_sales_growth_df = (
    monthly_sales_df
    .withColumn("prev_sales", F.lag("sales").over(window_sales))
    .withColumn("sales_growth", (F.col("sales") - F.col("prev_sales")) * 100 / F.col("prev_sales"))
    .orderBy("year_month")
)
display(monthly_sales_growth_df)
```

##Monthly Active Users

```python
monthly_customer_df = (
    year_month_df
    .groupBy("year_month")
    .agg(F.countDistinct("customer_id").alias("customers"))
    .orderBy("year_month")
)
display(monthly_customer_df)
```

## New and Existing Users

```python
# Find first order for each customers
# Drop the duplicate : we only want one customer for one month
unique_customers_df = year_month_df.dropDuplicates(["customer_id", "year_month"])

# Create a window to add a column "first order"
window_customers = Window.partitionBy("customer_id")

new_existing_users_df = (
    unique_customers_df
    .withColumn("first_invoice", F.min("year_month").over(window_customers))
    .groupBy("year_month")
    .agg(
        F.countDistinct("customer_id").alias("customers"),
        # Count the customer if the first invoice is the same than the current year_month. F.col("customer_id") manages null cases
        F.count(F.when(F.col("year_month") == F.col("first_invoice"), F.col("customer_id"))).alias("new_customers")
        )
    .withColumn("existing_customers", F.col("customers") - F.col("new_customers"))
    .orderBy("year_month")
)
display(new_existing_users_df)
```

## Finding RFM
RFM is a method used for analyzing customer value. It is commonly used in database marketing and direct marketing and has received particular attention in the retail and professional services industries. ([wikipedia](https://en.wikipedia.org/wiki/RFM_(market_research))

Optional Reading: [Making Your Database Pay Off Using Recency Frequency and Monetary Analysis](http://www.dbmarketing.com/2010/03/making-your-database-pay-off-using-recency-frequency-and-monetary-analysis/)

RFM stands for three dimensions:
- Recency – How recently did the customer purchase?
- Frequency – How often do they purchase?
- Monetary Value – How much do they spend?

Note: To simplify the problem, let's keep all placed and canceled orders.

```python
rfm_table_df = (
    invoice_line_amount_df
    .dropna(subset=["customer_id"])
    .groupby("customer_id")
    .agg(
        F.max(F.col("invoice_date")).alias("recency"),
        F.countDistinct(F.col("invoice_no")).alias("frequency"),
        F.sum(F.col("line_amount")).alias("monetary")
    )
    # Get the difference of days between last visit and today
    .withColumn("recency", F.datediff(F.current_date(), F.col("recency")))
    .orderBy("customer_id")
)

display(rfm_table_df)
```

### RFM Segmentation

##### RFM Score Values
To replace `pd.qcut`, we can use `QuantileDiscretizer` or `ntile` in PySpark:
- `QuantileDiscretizer` uses statistical quantiles like `qcut`: its behavior is virtually identical. You have an API with ML Pipelines and you can customize the number of buckets, but since it is driven by statistics, it is not stable between all executions depending on the approximations.
- With `ntile`, the size between groups is exactly the same (modulo 5). It sorts all the data, then assigns each row to a compartment. It performs better on big data, so we will use it in our case.

```python
w_recency = Window.partitionBy(F.lit(1)).orderBy(F.col("recency").desc()) # A lower recency means a client that bought recently
w_freq = Window.partitionBy(F.lit(1)).orderBy(F.col("frequency"))
w_monetary = Window.partitionBy(F.lit(1)).orderBy(F.col("monetary"))

rfm_score_df = (
    rfm_table_df
    .withColumn("recency_score", F.ntile(5).over(w_recency))
    .withColumn("frequency_score", F.ntile(5).over(w_freq))
    .withColumn("monetary_score", F.ntile(5).over(w_monetary))
    .withColumn("rfm_score", F.concat(F.col("recency_score"), F.col("frequency_score"), F.col("monetary_score")))
)
display(rfm_score_df)
```

Segmenting of customers according to Recency score and Frequency score values

```python
seg_map = {
    '[1-2][1-2]': 'Hibernating',
    '[1-2][3-4]': 'At Risk',
    '[1-2]5': 'Can\'t Lose',
    '3[1-2]': 'About to Sleep',
    '33': 'Need Attention',
    '[3-4][4-5]': 'Loyal Customers',
    '41': 'Promising',
    '51': 'New Customers',
    '[4-5][2-3]': 'Potential Loyalists',
    '5[4-5]': 'Champions'
}
```

```python
# Encapsulate a function to retrieve the segment name from the recency and frequency score
def map_segment(col, mapping):
    res = None
    for regex, seg in mapping.items():
        cond = col.rlike(f"^{regex}$")
        if res is None:
            res = F.when(cond, seg)
        else:
            res = res.when(cond, seg)
    return res.otherwise("Unknown")

rfm_segment_df = (
    rfm_score_df
    .withColumn("segment", F.concat("recency_score", "frequency_score"))
    .withColumn("segment", map_segment(F.col("segment"), seg_map))
)

display(rfm_segment_df)
```

```python
rfm_segment_stat_df = (
    rfm_segment_df
    .groupBy("segment")
    .agg(
        F.count(F.col("customer_id")).alias("# customers"),
        F.round(F.mean(F.col("recency"))).alias("avg_recency"),
        F.round(F.mean(F.col("frequency"))).alias("avg_frequency"),
        F.round(F.mean(F.col("monetary"))).alias("avg_monetary")
    )
    .orderBy("segment")
)
display(rfm_segment_stat_df)
```
