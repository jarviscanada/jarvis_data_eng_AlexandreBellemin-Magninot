# Silver to Gold process
For this part, we want to answer many questions:
- Which day(s) of the week sees the highest number of fraudulent transactions?
- What is the trend of the fraud rate (fraudulent transactions divided by total) over the past month?
- Which users have the largest number of flagged `is_fraud = true` transactions?
- Are there any users showing a sharp rise in transaction amount compared to their weekly average?
- Which merchant categories exhibit the highest fraud rate?
- Are there specific merchants with unusually high fraud volume?
- How does fraud distribution vary by time of day (morning vs night)?
- What’s the average transaction amount for fraud vs non-fraud transactions?
- Which merchant category has the highest total fraud amount?
- What are the total monetary losses due to fraud each day?
- How many unique users commit fraudulent transactions per week?
- Do fraud patterns show seasonal or monthly spikes?
- How has user behavior changed before versus after a fraudulent event?
- Are fraudulent transactions more common on high-value purchases compared to low-value purchases?

For each question, we will create a new table, which will be the source of our future dashboard.

```python
# Import the silver tables
from pyspark.sql import functions as F

transactions = spark.read.table('jarvis_tic.02_silver.transactions_data')
cards = spark.read.table('jarvis_tic.02_silver.cards_data')
users = spark.read.table('jarvis_tic.02_silver.users_data')
display(transactions)
display(cards)
display(users)
```

### Day with highest number of fraudulent transactions
Which day(s) of the week sees the highest number of fraudulent transactions?

```python
day_week_total_fraudulent_transactions = (
    transactions
    .withColumn('day_of_week', F.date_format('date', 'EEEE'))
    .groupBy('day_of_week')
    .agg(F.count_if(F.col('fraud')).alias('total_fraudulent_transactions'))
    .orderBy(F.col('total_fraudulent_transactions').desc())
)
day_week_total_fraudulent_transactions.write.mode('overwrite').saveAsTable('jarvis_tic.03_gold.day_week_highest_fraudulent_transactions')

display(day_week_total_fraudulent_transactions)
```

### Trend of fraud rate
What is the trend of the fraud rate (fraudulent transactions divided by total) over the past month?

```python
max_date = transactions.agg(F.max('date')).collect()[0]['max(date)']
display(max_date)

trend_fraud_rate_last_month = (
    transactions
    .filter(F.month(F.col('date')) == max_date.month)
    .filter(F.year(F.col('date')) == max_date.year)
    .agg(F.count(F.col('transaction_id')).alias('total_transactions'), 
         F.count_if(F.col('fraud')).alias('total_fraudulent_transactions'))
    .withColumn('fraud_rate', F.col('total_fraudulent_transactions') / F.col('total_transactions'))
    .withColumn('last_date', F.date_format(F.lit(max_date), "yyyy-MM"))
)
trend_fraud_rate_last_month.write.mode('overwrite').saveAsTable('jarvis_tic.03_gold.trend_fraud_rate_past_months')

display(trend_fraud_rate_last_month)
```

### Users with largest fraud transactions
Which users have the largest number of flagged (“is_fraud = true”) transactions?

```python
users_fraud_transactions = (
    transactions
    .groupBy("client_id")
    .agg(F.count_if(F.col('fraud')).alias('fraud_transactions'), F.count(F.col('transaction_id')).alias('total_transactions'))
    .sort(F.col('fraud_transactions').desc())
)
users_fraud_transactions.write.mode('overwrite').saveAsTable('jarvis_tic.03_gold.users_fraud_transactions')

display(users_fraud_transactions)
```

### Users with rise in transaction
Are there any users showing a sharp rise in transaction amount compared to their weekly average?

```python
from pyspark.sql.window import Window
from pyspark.sql.types import DecimalType

transactions_year_week = (
    transactions
    .withColumn("year", F.year("date"))
    .withColumn("week", F.weekofyear("date"))
)

# windwow 1: avg per client / week
w_week = Window.partitionBy("client_id", "year", "week")

# window 2: avg per client
w_client = Window.partitionBy("client_id")

sharp_rise_users_transactions = (
    transactions_year_week
    .withColumn(
        "avg_transactions_amount",
        F.avg("amount").over(w_week)
        .cast(DecimalType(18,2))
    )
    # Calculate the avg / week
    .withColumn(
        "avg_week_amount",
        F.avg("avg_transactions_amount").over(w_client)
        .cast(DecimalType(18,2))
    )
    # Calculate the std / week
    .withColumn(
        "std_week_amount",
        F.stddev("avg_transactions_amount").over(w_client)
    )
    .drop("avg_transactions_amount")
    # Calculate the z score
    .withColumn(
        "z_score",
        (F.col("amount") - F.col("avg_week_amount")) / F.col("std_week_amount")
    )
    # We determine here that a sharp rise is 2 times or more than the z score
    .filter(F.col("z_score") > 2)
    .select("client_id", "date", "amount", "z_score", "fraud")
)
sharp_rise_users = (
    sharp_rise_users_transactions
    .groupBy("client_id")
    .agg(F.count(F.col('client_id')).alias('count_sharp_rises'))
)
display(sharp_rise_users_transactions)
display(sharp_rise_users)

# We keep the both tables
sharp_rise_users.write.mode('overwrite').saveAsTable('jarvis_tic.03_gold.sharp_rise_users')
sharp_rise_users_transactions.write.mode('overwrite').saveAsTable('jarvis_tic.03_gold.sharp_rise_users_transactions')
```

### Merchant with highest fraud rate
Which merchant categories exhibit the highest fraud rate?

```python
merchants_cat_fraud_volume = (
    transactions
    .groupBy('mcc_label')
    .agg(F.count_if(F.col('fraud')).alias('total_fraud_transactions'), 
         F.count('transaction_id').alias('total_transactions'))
    .withColumn('fraud_rate', F.col('total_fraud_transactions')/F.col('total_transactions'))
    .sort(F.desc('fraud_rate'))
)
merchants_cat_fraud_volume.write.mode('overwrite').saveAsTable('jarvis_tic.03_gold.merchants_cat_fraud_volume')

display(merchants_cat_fraud_volume)
```

### Merchants with high fraud volume
Are there specific merchants with unusually high fraud volume?

```python
merchants_fraud_volume = (
    transactions
    .groupBy('merchant_id')
    .agg(F.count_if(F.col('fraud')).alias('total_fraud_transactions'), 
         F.count('transaction_id').alias('total_transactions'))
    .withColumn('fraud_rate', F.col('total_fraud_transactions')/F.col('total_transactions'))
    .sort(F.desc('total_fraud_transactions'))
)
merchants_fraud_volume.write.mode('overwrite').saveAsTable('jarvis_tic.03_gold.merchants_fraud_volume')

display(merchants_fraud_volume)
```

### Fraud distribution by time of day
How does fraud distribution vary by time of day (morning vs night)?

```python
fraud_distribution_time_day = (
    transactions
    .filter(F.col('fraud') == True)
    .withColumn('transaction_hour', F.date_format(F.col('date'), 'HH'))
    .groupBy('transaction_hour')
    .agg(F.count(F.col('transaction_id')).alias('total_fraud_transactions'))
    .withColumn('am_pm', 
                    F.when(F.col('transaction_hour') < 12, 'AM')
                    .when(F.col('transaction_hour') >= 12, 'PM')
    )
    .sort('transaction_hour')
)
fraud_distribution_time_day.write.mode('overwrite').saveAsTable('jarvis_tic.03_gold.fraud_distribution_time_day')

display(fraud_distribution_time_day)
```

### Average transaction amount for fraud and non-fraud
What’s the average transaction amount for fraud vs non-fraud transactions?

```python
avg_transactions_fraud_or_not = (
    transactions
    .groupBy('fraud')
    .agg(F.avg(F.col('amount')).alias('avg_transactions_amount'))
)
avg_transactions_fraud_or_not.write.mode('overwrite').saveAsTable('jarvis_tic.03_gold.avg_transactions_fraud_or_not')

display(avg_transactions_fraud_or_not)
```

### Merchant category with the highest fraud amount
Which merchant category has the highest total fraud amount?

```python
merchant_cat_fraud_amount = (
  transactions
  .filter(F.col('fraud'))
  .groupBy('mcc_label')
  .agg(F.sum(F.col('amount')).alias('total_fraud_amount'))
  .sort(F.col('total_fraud_amount').desc())
)
merchant_cat_fraud_amount.write.mode('overwrite').saveAsTable('jarvis_tic.03_gold.merchant_cat_fraud_amount')

display(merchant_cat_fraud_amount)
```

### Total monetary losses per day
What are the total monetary losses due to fraud each day?

```python
total_monetary_loss_each_day = (
    transactions
    .filter(F.col('fraud'))
    .withColumn('day' , F.date_format(F.col('date'), 'yyyy-MM-dd'))
    .groupBy('day')
    .agg(F.sum(F.col('amount')).alias('total_monetary_loss'))
    .sort(F.col('day'))
)
total_monetary_loss_each_day.write.mode('overwrite').saveAsTable('jarvis_tic.03_gold.total_monetary_loss_each_day')

display(total_monetary_loss_each_day)
```

### Unique users fraud per week
How many unique users commit fraudulent transactions per week?

```python
unique_users_fraud_per_week = (
    transactions
    .filter(F.col('fraud'))
    .withColumn('year', F.date_format(F.col('date'), 'yyyy'))
    .withColumn('week', F.weekofyear(F.col('date')))
    .groupBy('year', 'week')
    .agg(F.countDistinct(F.col('client_id')).alias('unique_users_fraud'))
    .sort(F.col('year'), F.col('week'))
)
unique_users_fraud_per_week.write.mode('overwrite').saveAsTable('jarvis_tic.03_gold.unique_users_fraud_per_week')

display(unique_users_fraud_per_week)
```

### Fraud patterns spikes
Do fraud patterns show seasonal or monthly spikes?

```python
fraud_patterns_year_month = (
    transactions
    .filter(F.col('fraud'))
    .withColumn('year_month', F.date_format(F.col('date'), 'yyyy-MM'))
    .groupBy('year_month')
    .agg(F.count(F.col('transaction_id')).alias('total_fraud_transactions'))
    .sort(F.col('year_month'))
)
fraud_patterns_year_month.write.mode('overwrite').saveAsTable('jarvis_tic.03_gold.fraud_patterns_year_month')

display(fraud_patterns_year_month)
```

### User behavior after a fraudulent event
How has user behavior changed before versus after a fraudulent event?

```python
from pyspark.sql.window import Window
from pyspark.sql.types import DecimalType

# Get first fraud of each user
w_client = Window.partitionBy("client_id")

behavior_after_fraudulent_event = (
    transactions
    .withColumn(
        "first_fraud_date",
        F.min(F.when(F.col("fraud"), F.col("date"))).over(w_client)
    )
    # Determine if the transactions is before or after the first fraud
    .withColumn(
        "period",
        F.when(F.col("first_fraud_date").isNull(), "no_fraud")
        .when(F.col("date") < F.col("first_fraud_date"), "before")
        .otherwise("after")
    )
    # We don't want users w/out fraud
    .filter(F.col("first_fraud_date").isNotNull())
    .groupBy("client_id", "period")
    # Get the number of transactions, total amount, avg amount
    # IDEA: get frequency
    .agg(
        F.count(F.col("transaction_id")).alias("number_of_transactions"),
        F.sum(F.col("amount")).alias("total_amount"),
        F.avg(F.col("amount")).alias("avg_amount").cast(DecimalType(18,2)),
        F.min(F.col("date")).alias("first_transaction_date"),
        F.max(F.col("date")).alias("last_transaction_date"),
        F.min(F.col("first_fraud_date")).alias("first_fraud_date")
    )
)
display(behavior_after_fraudulent_event)

behavior_summary = (
    behavior_after_fraudulent_event
    .groupBy("period")
    .agg(
        F.count(F.col("client_id")).alias("number_of_users"),
        F.sum(F.col("number_of_transactions")).alias("total_transactions"),
        F.sum(F.col("total_amount")).alias("total_amount"),
        F.avg(F.col("avg_amount")).alias("avg_amount").cast(DecimalType(18,2))
    )
)
display(behavior_summary)

behavior_summary.write.mode('overwrite').saveAsTable('jarvis_tic.03_gold.behavior_summary')
behavior_after_fraudulent_event.write.mode('overwrite').saveAsTable('jarvis_tic.03_gold.behavior_after_fraudulent_event')
```

### Fraud high vs low value purchases
Are fraudulent transactions more common on high-value purchases compared to low-value purchases?

```python
from pyspark.sql.window import Window

# Define limits between high and low value purchases
fraud_high_low_purchases = (
    transactions
    .withColumn(
        "quantile",
        F.ntile(4).over(Window.partitionBy().orderBy(F.col("amount")))
    )
    # Calculate fraud rate for each quantile
    .groupBy("quantile")
    .agg(
        F.count(F.col("transaction_id")).alias("total_transactions"),
        F.count_if(F.col("fraud")).alias("total_fraud_transactions")
    )
    .withColumn(
        "fraud_rate",
        F.col("total_fraud_transactions") / F.col("total_transactions")
    )
)
display(fraud_high_low_purchases)

fraud_high_low_purchases.write.mode('overwrite').saveAsTable('jarvis_tic.03_gold.fraud_high_low_purchases')
```
